### 📌 Objectives:
1. Load and inspect the preprocessed dataset (`student_performance_prepared.csv`).
2. Define feature matrix ($X$) and target vector ($y$).
3. Perform train-test split (80% training, 20% testing) with reproducibility seed (`random_state=42`).
4. Fit an ordinary least squares Linear Regression model.
5. Compute performance evaluation metrics: Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and $R^2$ Score.
6. Visualize Actual vs. Predicted values with an ideal reference line.
7. Perform inference on new student profiles.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load prepared dataset
dataset = pd.read_csv("student_performance_prepared.csv")
print("Dataset size:", dataset.shape)
display(dataset.head())
```

```python
print("Column names:")
print(dataset.columns.tolist())

print("\nData types:")
display(dataset.dtypes.to_frame("Data Type"))
```

```python
# Feature Matrix (X) and Target Vector (y)
X = dataset.drop(columns=["Final_Exam_Score"])
y = dataset["Final_Exam_Score"]

print("Input features count:", X.shape[1])
print("Target variable: Final_Exam_Score")
print("Features:", X.columns.tolist())
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(dataset["Study_Hours_Per_Week"], dataset["Final_Exam_Score"], alpha=0.5, color="#1f77b4")
plt.title("Study Hours (Scaled) vs. Final Exam Score")
plt.xlabel("Standardized Study Hours")
plt.ylabel("Final Exam Score")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Split dataset into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))
```

```python
# Initialize and train Linear Regression model
reg = LinearRegression()
reg.fit(X_train, y_train)

print(f"Model Intercept (bias): {reg.intercept_:.4f}")
coefficients_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": reg.coef_
}).sort_values(by="Coefficient", ascending=False)

print("\n--- Model Coefficients (Feature Weights) ---")
display(coefficients_df)
```

```python
# Predict on test set
predicted = reg.predict(X_test)

results = pd.DataFrame({
    "Actual Score": y_test.values,
    "Predicted Score": np.round(predicted, 2),
    "Residual (Error)": np.round(y_test.values - predicted, 2)
})

print("--- Actual vs. Predicted Sample ---")
display(results.head(10))
```

```python
# Compute Evaluation Metrics
mae = mean_absolute_error(y_test, predicted)
mse = mean_squared_error(y_test, predicted)
rmse = mse ** 0.5
r2 = r2_score(y_test, predicted)

print("=" * 45)
print("📊 MODEL EVALUATION METRICS")
print("=" * 45)
print(f"Mean Absolute Error (MAE)  : {mae:.3f}")
print(f"Mean Squared Error (MSE)   : {mse:.3f}")
print(f"Root Mean Squared Error    : {rmse:.3f}")
print(f"R-squared (R² Score)       : {r2:.4f} ({r2 * 100:.2f}% Variance Explained)")
print("=" * 45)
```

```python
# Scatter plot: Actual vs Predicted
plt.figure(figsize=(6, 6))
plt.scatter(y_test, predicted, alpha=0.6, color="#2e7d32", edgecolors="k")
low = min(y_test.min(), predicted.min())
high = max(y_test.max(), predicted.max())
plt.plot([low, high], [low, high], "--", color="red", linewidth=2, label="Ideal Fit (y = x)")
plt.title("Actual vs. Predicted Final Exam Scores", fontsize=12, fontweight="bold")
plt.xlabel("Actual Score")
plt.ylabel("Predicted Score")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Inference on a hypothetical new student
sample_student = X_test.iloc[[0]].copy()
new_pred = reg.predict(sample_student)[0]

print("--- Hypothetical Student Features ---")
display(sample_student)
print(f"\n🎯 Predicted Final Exam Score: {new_pred:.2f} (Actual was: {y_test.iloc[0]:.2f})")
```
